```python
from langgraph.graph import StateGraph, END
from typing import TypedDict
import re
```

```python
# 📌 Step 1: Define TypedDict State Structure
class EmailState(TypedDict):
    email: str
    is_valid: bool
    message: str
```

```python
# 📌 Step 2: Define Nodes (Agents)
def input_email(state: EmailState) -> EmailState:
    print("Received Email:", state["email"])
    return state
```

```python
def validate_email(state: EmailState) -> EmailState:
    pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
    if re.match(pattern, state["email"]):
        print("Valid Email")
        state["is_valid"] = True
        state["message"] = "Welcome! Your email is verified."
    else:
        print("Invalid Email")
        state["is_valid"] = False
        state["message"] = "Error: Invalid email format."
    return state
```

```python
def end_node(state: EmailState) -> EmailState:
    print("Final Message:", state["message"])
    return state
```

```python
# 📌 Step 3: Build Graph
builder = StateGraph(EmailState)
```

```python
builder.add_node("input_email", input_email)
builder.add_node("validate_email", validate_email)
builder.add_node("end", end_node)
```

```python
builder.set_entry_point("input_email")
builder.add_edge("input_email", "validate_email")
builder.add_edge("validate_email", "end")
builder.add_edge("end", END)
```

```python
graph = builder.compile()
```

```python
# 📌 Step 4: Test With User Input
test_input = {
    "email": "emon@example.com",
    "is_valid": False,
    "message": ""
}
```

```python
final_state = graph.invoke(test_input)
```

```python
print("\n✅ Final State:", final_state)
```
