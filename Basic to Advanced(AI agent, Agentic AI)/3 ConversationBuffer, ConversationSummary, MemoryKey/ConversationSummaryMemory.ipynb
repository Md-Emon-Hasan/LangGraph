```python
# import depencncy
import os
from langgraph.graph import StateGraph
from langgraph.graph import END
from typing import TypedDict

from langchain.chains import ConversationChain
from langchain.memory import ConversationSummaryMemory
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
# load api key
from dotenv import load_dotenv

load_dotenv()
```

```python
# Define Shared State
class ChatState(TypedDict):
    input: str
    output: str
```

```python
# Gemini LLM
llm = ChatGoogleGenerativeAI(
    model = 'gemini-2.5-pro',
    temperature=0.9,
    convert_system_message_to_human=True
)
```

```python
# setup conversatinSummaryMemory
summary_memory = ConversationSummaryMemory(
    llm=llm,
    memory_key='history',
    return_messages=True
)
```

```python
# Conversation Chain with summary memory
conversation = ConversationChain(
    llm=llm,
    memory=summary_memory
)
```

```python
# Define LangGraph Node
def chat_node(state: ChatState) -> ChatState:
    user_input = state["input"]
    print("User:", user_input)

    response = conversation.predict(input=user_input)
    print("Gemini:", response)

    state["output"] = response
    return state
```

```python
# langGraph build
builder = StateGraph(ChatState)

builder.add_node('chat',chat_node)
builder.set_entry_point('chat')
builder.add_edge('chat',END)
```

```python
graph = builder.compile()
```

```python
# Multi-Turn Chat with Summary Memory
print("\nTurn 1")
state1 = graph.invoke({"input": "Hi Gemini, I’m planning a trip to Japan.", "output": ""})
```

```python
print("\nTurn 2")
state2 = graph.invoke({"input": "Suggest me 3 must-visit places.", "output": ""})
```

```python
print("\nTurn 3")
state3 = graph.invoke({"input": "What did I tell you about my trip?", "output": ""})
```
