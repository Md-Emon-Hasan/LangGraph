```python
from langgraph.graph import StateGraph, END
from typing import TypedDict
```

```python
# 📌 Step 1: Define shared state
class UserState(TypedDict):
    name: str
    valid: bool
    greeting: str
```

```python
# 📌 Step 2: Define node functions
def start_node(state: UserState) -> UserState:
    print("Received input name:", state["name"])
    return state
```

```python
def validate_user(state: UserState) -> UserState:
    if state["name"] and state["name"].strip():
        print("Name is valid")
        state["valid"] = True
    else:
        print("Invalid name detected")
        state["valid"] = False
        state["greeting"] = "Error: Name cannot be empty."
    return state
```

```python
def greet_user(state: UserState) -> UserState:
    if state["valid"]:
        state["greeting"] = f"Hello, {state['name']}! Welcome to LangGraph"
    return state
```

```python
def end_node(state: UserState) -> UserState:
    print("Final Output:", state["greeting"])
    return state
```

```python
# 📌 Step 3: Build the LangGraph
builder = StateGraph(UserState)

builder.add_node("start", start_node)
builder.add_node("validate_user", validate_user)
builder.add_node("greet_user", greet_user)
builder.add_node("end", end_node)
```

```python
# 📌 Step 4: Define flow
builder.set_entry_point("start")
builder.add_edge("start", "validate_user")
builder.add_edge("validate_user", "greet_user")
builder.add_edge("greet_user", "end")
builder.add_edge("end", END)
```

```python
# 📌 Step 5: Compile the graph
graph = builder.compile()
```

```python
# 📌 Step 6: Run graph with input
user_input = {"name": "Emon", "valid": False, "greeting": ""}
final_state = graph.invoke(user_input)
```

```python
# ✅ Final output
print("\nFinal State:", final_state)
```
